FLATTEN PER PREPARARE A DUCKDB 
i dati su OpenFDA sono JSON annidati, quindi avere per esempio 3 farmaci e 5 reazioni in un un json unico. bisogna predisporre per SQL (duckdb probabilmente) trasformando in tabelle piatte (matrici semplici dove ogni riga è un'osservazione)

```python
import zipfile
import ijson
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq
from pathlib import Path

# Percorsi
RAW_DIR = Path("../data/raw")
PARQUET_DIR = Path("../data/parquet")
PARQUET_DIR.mkdir(parents=True, exist_ok=True)

# File già scaricati
files = sorted(RAW_DIR.glob("*.json.zip"))
print(f"File disponibili: {len(files)}")
for f in files:
    print(f"  {f.name}")
```

```python
def parse_age_to_years(age, unit):
    if age is None:
        return None
    try:
        age = float(age)
        unit_map = {
            '800': age / 10,
            '801': age,
            '802': age / 12,
            '803': age / 52,
            '804': age / 365,
            '805': age / 8760,
        }
        return round(unit_map.get(str(unit), age), 2)
    except:
        return None

def age_to_group(age_years):
    if age_years is None:
        return None
    if age_years < 2:   return 'neonate'
    if age_years < 12:  return 'pediatric'
    if age_years < 18:  return 'adolescent'
    if age_years < 65:  return 'adult'
    return 'geriatric'

def get_drug_name(drug):
    active = drug.get('activesubstance', {})
    if active and active.get('activesubstancename'):
        return active['activesubstancename'].upper().strip()
    openfda = drug.get('openfda', {})
    if openfda and openfda.get('generic_name'):
        return openfda['generic_name'][0].upper().strip()
    mp = drug.get('medicinalproduct', '')
    return mp.upper().strip() if mp else None

def get_pharm_class(drug):
    openfda = drug.get('openfda', {})
    if openfda and openfda.get('pharm_class_epc'):
        return openfda['pharm_class_epc'][0]
    return None
```

_flatten_file_ -> apre il json.zip, itera sui record uno alla volta senza caricare in ram (con _ijson_) e crea una riga per report, con tutti i campi "piatti", tipo ID del report, data, paese, seriousness, sesso, age, agegroup. poi estende l'array dei farmaci, con N righe per drugs. una riga per ogni drug del report, cercando il nome secondo la logica di priorita (active substance -> openfda generic name -> medicinalproduct. estende poi anche le reactions, creando una riga per ogni reazione. restituisce quindi 3 tabelle collegate dal report ID. 

```python
def flatten_file(zip_path):
    reports, drugs, reactions = [], [], []
    
    with zipfile.ZipFile(zip_path) as zf:
        json_file = [f for f in zf.namelist() if f.endswith('.json')][0]
        with zf.open(json_file) as f:
            for i, report in enumerate(ijson.items(f, 'results.item')):
                try:
                    rid = report.get('safetyreportid')
                    patient = report.get('patient', {}) or {}
                    age_years = parse_age_to_years(
                        patient.get('patientonsetage'),
                        patient.get('patientonsetageunit')
                    )

                    reports.append({
                        'report_id':            rid,
                        'receiptdate':          report.get('receiptdate'),
                        'receivedate':          report.get('receivedate'),
                        'occurcountry':         report.get('occurcountry'),
                        'primarysourcecountry': report.get('primarysourcecountry'),
                        'serious':              report.get('serious'),
                        'death':                report.get('seriousnessdeath'),
                        'hospitalization':      report.get('seriousnesshospitalization'),
                        'lifethreatening':      report.get('seriousnesslifethreatening'),
                        'disabling':            report.get('seriousnessdisabling'),
                        'sex':                  patient.get('patientsex'),
                        'age_years':            age_years,
                        'age_group':            age_to_group(age_years),
                        'weight_kg':            patient.get('patientweight'),
                        'reporter_qualification': (report.get('primarysource') or {}).get('qualification'),
                    })

                    for j, drug in enumerate(patient.get('drug', []) or []):
                        if drug is None:
                            continue
                        drugs.append({
                            'report_id':        rid,
                            'drug_seq':         j,
                            'drug_name':        get_drug_name(drug),
                            'medicinalproduct': drug.get('medicinalproduct'),
                            'characterization': drug.get('drugcharacterization'),
                            'indication':       drug.get('drugindication'),
                            'route':            drug.get('drugadministrationroute'),
                            'dosage_form':      drug.get('drugdosageform'),
                            'pharm_class_epc':  get_pharm_class(drug),
                            'actiondrug':       drug.get('actiondrug'),
                        })

                    for j, rxn in enumerate(patient.get('reaction', []) or []):
                        if rxn is None:
                            continue
                        reactions.append({
                            'report_id':        rid,
                            'reaction_seq':     j,
                            'reaction_pt':      rxn.get('reactionmeddrapt'),
                            'reaction_outcome': rxn.get('reactionoutcome'),
                            'meddra_version':   rxn.get('reactionmeddraversionpt'),
                        })

                except Exception as e:
                    import traceback, json
                    print(f"Errore al record {i}, id: {report.get('safetyreportid')}")
                    print(f"  Tipo errore: {type(e).__name__}: {e}")
                    traceback.print_exc()
                    print("\nRecord:")
                    print(json.dumps(report, indent=2))
                    break

    return pd.DataFrame(reports), pd.DataFrame(drugs), pd.DataFrame(reactions)

# Test
df_rep, df_drug, df_rxn = flatten_file(files[0])
print(f"Reports processati: {len(df_rep)}")
```

```python
for zip_file in files:
    print(f"\n{'='*50}")
    print(f"Elaborando: {zip_file.name}")
    
    try:
        df_rep, df_drug, df_rxn = flatten_file(zip_file)
        
        print(f"  Reports:   {len(df_rep):>6}")
        print(f"  Drugs:     {len(df_drug):>6}")
        print(f"  Reactions: {len(df_rxn):>6}")
        print(f"  Drug name null:       {df_drug['drug_name'].isna().sum():>4} ({df_drug['drug_name'].isna().mean():.1%})")
        print(f"  Age group coverage:   {df_rep['age_group'].notna().mean():.1%}")
        print(f"  Reaction outcome:     {df_rxn['reaction_outcome'].notna().mean():.1%}")
        print(f"  Pharm class coverage: {df_drug['pharm_class_epc'].notna().mean():.1%}")
        
    except Exception as e:
        print(f"  ERRORE: {e}")
```
